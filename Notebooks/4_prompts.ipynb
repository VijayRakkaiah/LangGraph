### **Prompts**

```python
import os
from dotenv import load_dotenv

if os.environ.get("OPENAI_API_KEY"):
    print("OPENAI_API_KEY is available")
else:
    raise ValueError("OPENAI_API_KEY is not available")

print()

from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model = "gpt-5.4-nano",
    temperature = 0.7
)

llm
```

```python
from langchain_core.prompts import ChatPromptTemplate

my_prompts = ChatPromptTemplate([
    ("system", "You are a helpful assistant that helps to create curate the post for social media platform"),
    ("human", "i want to create the post about the importance of data privacy in the digital age")
])
```

```python
my_prompts
```

```python
my_prompts_1 = ChatPromptTemplate([
    ("system", "You are a helpful assistant that helps to create curate the post for social media platform"),
    ("human", "I want to create a post about {topic}")
])
```

```python
my_prompts_1.invoke({"topic":"The importance of data privacy in the digital age"})
```

```python
filled_prompt = my_prompts_1.invoke({"topic":"The importance of data privacy in the digital age"})

llm.invoke(filled_prompt)
```

### **Chain with Prompts**

```python
chain = my_prompts_1 | llm

chain
```

```python
chain.invoke({"topic": "The importance of the data privacy in digital age"})
```
