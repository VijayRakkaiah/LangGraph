```python
import os
from dotenv import load_dotenv

if os.environ.get("OPENAI_API_KEY"):
    print("OPENAI_API_KEY is available")
else:
    raise ValueError("OPENAI_API_KEY is not available")

print()

from langchain_openai import ChatOpenAI

llm = ChatOpenAI(
    model = "gpt-5.4-nano",
    temperature = 0.7
)

llm
```

## **Tools**

### **Duck Duck Go Search Tool**

```python
from langchain_community.tools import DuckDuckGoSearchRun
from langchain.tools import tool

@tool
def search_duckduckgo(query: str) -> str:

    """This tool searches the latest news on DuckDuckGo for the given query and return the results."""

    search = DuckDuckGoSearchRun()

    return search.invoke(query)

search_duckduckgo.invoke("what is the latest news about langchain and langgraph")
```

### **Arxiv Query Tool**

```python
from langchain_community.tools import ArxivQueryRun
from langchain_community.utilities import ArxivAPIWrapper

@tool
def arxiv_tool(query: str) -> str:

    """This tool allows you to query the arXiv database for research papers."""

    arxiv_query = ArxivQueryRun(api_wrapper=ArxivAPIWrapper())

    return arxiv_query.invoke(query)

arxiv_tool.invoke("Transformer model in NLP")
```

### **Wikipedia Search Tool**

```python
from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper

@tool
def wiki_tool(query: str) -> str:

    """This tool allows you to search wikipedia for information on a given topic."""

    wikipedia_search = WikipediaQueryRun(api_wrapper=WikipediaAPIWrapper())

    return wikipedia_search.invoke(query)

wiki_tool.invoke("langgraph")
```

### **Custom Tool**

```python
@tool
def custom_tool(name: str):
    """use this tool to get personal information about Vijay, Raman or Mahathi"""

    info = {
        "Vijay": "Vijay is a GenAI developer with 5 years of experience in AI",
        "Raman": "Raman is a software engineer with 5 years experience in AIML",
        "Mahathi": "Mahathi is product manager with a background in tech startup"
    }

    return info.get(name, "No information available for this person")
```

```python
custom_tool.invoke('Vijay')
```

## **Tool Binding**

```python
tools = [search_duckduckgo, arxiv_tool, wiki_tool, custom_tool]

llm_with_tools = llm.bind_tools(tools)
```

```python
response = llm_with_tools.invoke("give me the information about Vijay use custom tool")
response
```
